```python
import pandas as pd
```

```python
df = pd.read_csv("Hotel_Reservation_Cleaned.csv")
```

```python
df.head()
```

```python
# Total revenue (using ADR - Average Daily Rate)
total_revenue = df["adr"].sum()

# Average room price
average_price = df["adr"].mean()

# Minimum room price
minimum_price = df["adr"].min()

# Maximum room price
maximum_price = df["adr"].max()

# Total number of bookings
total_bookings = df["adr"].count()

print("===== DATA ANALYSIS =====")
print(f"Total Revenue (ADR Sum): {total_revenue:.2f}")
print(f"Average Room Price: {average_price:.2f}")
print(f"Minimum Room Price: {minimum_price:.2f}")
print(f"Maximum Room Price: {maximum_price:.2f}")
print(f"Total Bookings: {total_bookings}")
```

```python
# Additional Business Insights
print("\n===== BUSINESS INSIGHTS =====")
print("Average Adults per Booking:", df["adults"].mean())
print("Average Children per Booking:", df["children"].mean())
print("Most Common Customer Type:")
print(df["customer_type"].value_counts().head(1))
print("\nMost Bookings by Hotel:")
print(df["hotel"].value_counts())
print("\nMost Bookings by Month:")
print(df["arrival_date_month"].value_counts().head())
```

```python
# Save analysis
analysis = pd.DataFrame({
    "Metric": [
        "Total Revenue (ADR Sum)",
        "Average Room Price",
        "Minimum Room Price",
        "Maximum Room Price",
        "Total Bookings"
    ],
    "Value": [
        total_revenue,
        average_price,
        minimum_price,
        maximum_price,
        total_bookings
    ]
})

analysis.to_csv("Hotel_Data_Analysis.csv", index=False)

print("\nAnalysis saved successfully!")
```
