```python
import numpy as np

# 1D Array
arr1 = np.array([10, 20, 30, 40, 50])
print(arr1)

# 2D Array
arr2 = np.array([[1, 2, 3],
                 [4, 5, 6]])
print(arr2)
```

```python
# Array Properties

import numpy as np

arr = np.array([[1, 2, 3],
                [4, 5, 6]])

print("Shape:", arr.shape)
print("Dimensions:", arr.ndim)
print("Size:", arr.size)
print("Data Type:", arr.dtype)
```

```python
# Mathematical Operations

import numpy as np

a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

```python
# Scalar Operations

import numpy as np

arr = np.array([2, 4, 6, 8])

print(arr + 5)
print(arr - 2)
print(arr * 3)
print(arr / 2)
```

```python
# Aggregate Functions

import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Sum:", np.sum(arr))
print("Average:", np.mean(arr))
print("Maximum:", np.max(arr))
print("Minimum:", np.min(arr))
print("Standard Deviation:", np.std(arr))
```

```python
# Array reshaping

import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6])

new_arr = arr.reshape(2, 3)

print(new_arr)
```

```python
# Array indexing and Slicing

import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print(arr[0])
print(arr[2])
print(arr[1:4])
```
