```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load dataset
df = pd.read_csv("Hotel_Reservation_Cleaned.csv")
```

```python
# 1. Bar Chart
# Hotel Booking Count
# -----------------------------
hotel_counts = df["hotel"].value_counts()

plt.figure(figsize=(8,4))
hotel_counts.plot(kind="bar")
plt.title("Number of Bookings by Hotel")
plt.xlabel("Hotel")
plt.ylabel("Number of Bookings")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# 2. Pie Chart
# Reservation Status Distribution
# -----------------------------
status_counts = df["reservation_status"].value_counts()

plt.figure(figsize=(4,4))
plt.pie(status_counts,
        labels=status_counts.index,
        autopct='%1.1f%%',
        startangle=90)
plt.title("Reservation Status Distribution")
plt.axis('equal')
plt.show()
```

```python
# Average Daily Rate by Room Type (Bar Chart)
room_price = df.groupby("reserved_room_type")["adr"].mean().sort_values()

plt.figure(figsize=(4,4))
room_price.plot(kind="bar")
plt.title("Average Daily Rate by Room Type")
plt.xlabel("Room Type")
plt.ylabel("Average Daily Rate")
plt.tight_layout()
plt.show()
```

```python
# Monthly Booking Trend (Line Chart)
month_order = ["January","February","March","April","May","June",
               "July","August","September","October","November","December"]

monthly = df["arrival_date_month"].value_counts().reindex(month_order)

plt.figure(figsize=(4,4))
plt.plot(monthly.index, monthly.values, marker='o')
plt.title("Monthly Booking Trend")
plt.xlabel("Month")
plt.ylabel("Bookings")
plt.xticks(rotation=90)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Market Segment Analysis (Horizontal Bar Chart)
market = df["market_segment"].value_counts()

plt.figure(figsize=(4,4))
market.sort_values().plot(kind="barh")

plt.title("Bookings by Market Segment")
plt.xlabel("Number of Bookings")
plt.ylabel("Market Segment")
plt.tight_layout()
plt.show()
```

```python
# Deposit Type Distribution (Pie Chart)
deposit = df["deposit_type"].value_counts()

plt.figure(figsize=(4,4))
plt.pie(deposit.values,
        labels=deposit.index,
        autopct="%1.1f%%",
        startangle=90)

plt.title("Deposit Type Distribution")
plt.axis("equal")
plt.show()
```

```python
# Lead Time by Hotel (Bar Chart)
lead = df.groupby("hotel")["lead_time"].mean()

plt.figure(figsize=(8,5))
lead.plot(kind="bar")

plt.title("Average Lead Time by Hotel")
plt.xlabel("Hotel")
plt.ylabel("Average Lead Time (Days)")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# Special Requests by Hotel (Bar Chart)
requests = df.groupby("hotel")["total_of_special_requests"].mean()

plt.figure(figsize=(7,5))
requests.plot(kind="bar")

plt.title("Average Special Requests by Hotel")
plt.xlabel("Hotel")
plt.ylabel("Average Special Requests")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```
