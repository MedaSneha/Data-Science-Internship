```python
# Import pandas
import pandas as pd

# Load the dataset
df = pd.read_csv("C:\\Users\\Pavan\\OneDrive\\Desktop\\Data Science Internship\\Hotel Booking Reservation Dataset.csv") 

# Display first 5 rows
df.head()
```

```python
# Checking dataset
print(df.isnull().sum())
```

```python
# Clean the dataset
df = df.drop_duplicates()
```

```python
# Fill missing numeric values with the median:

numeric_cols = df.select_dtypes(include=['number']).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
# Verify Cleaning
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
# Export the Cleaned Dataset
df.to_csv("Cleaned_Hotel_Booking_Reservation_Dataset.csv", index=False)

print("Cleaned dataset exported successfully!")
```
