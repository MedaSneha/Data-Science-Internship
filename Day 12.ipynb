```python
import pandas as pd
```

```python
# Load the hotel booking reservation dataset
df = pd.read_csv("C:\\Users\\Pavan\\OneDrive\\Desktop\\Data Science Internship\\Hotel Booking Reservation Dataset.csv") 

# Display the first five records
df.head()
```

```python
## For cleaning:

# Remove duplicate records from the dataset
df = df.drop_duplicates()

# Identify numerical columns
numeric_cols = df.select_dtypes(include=['number']).columns

# Fill missing numerical values with the median
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())
```

## 📊 Dataset Overview

```python
# Display dataset dimensions
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

# Display column names
print("\nColumns:")
print(df.columns.tolist())

# Display data types
print("\nData Types:")
print(df.dtypes)
```

### Room Type Distribution

```python
import matplotlib.pyplot as plt

# Count reservations by room type
room_counts = df['reserved_room_type'].value_counts()

plt.figure(figsize=(8, 5))
room_counts.plot(kind='bar')

plt.title('Reservations by Room Type')
plt.xlabel('Room Type')
plt.ylabel('Number of Reservations')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Booking Status Distribution

```python
import matplotlib.pyplot as plt

# Count booking status
booking_status = df['reservation_status'].value_counts()

# Create bar chart
plt.figure(figsize=(8, 5))
booking_status.plot(kind='bar')

plt.title('Booking Status Distribution')
plt.xlabel('Booking Status')
plt.ylabel('Number of Bookings')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 💡 Business Insights

1. The analysis shows differences in customer preferences across room types.

3. The number of rooms booked indicates that individual and small-group reservations are common.
