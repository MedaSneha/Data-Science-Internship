```python
import pandas as pd
```

```python
df = pd.read_csv("C:\\Users\\Pavan\\OneDrive\\Desktop\\Data Science Internship\\Hotel Booking Reservation Dataset.csv")
```

```python
print(df.head())
print(df.info())
print(df.shape)
```

```python
# check missing values
print(df.isnull().sum())
```

```python
# Total misiing values
print(df.isnull().sum().sum())
```

```python
# Filling missing values
df['country'] = df['country'].fillna(df['country'].mode()[0])
```

```python
df['agent'] = df['agent'].fillna(0)
df['company'] = df['company'].fillna(0)
```

```python
print(df[['country', 'agent', 'company']].isnull().sum())
```

```python
print("Duplicate Rows:", df.duplicated().sum())
```

```python
print("Duplicate Rows After Removing:", df.duplicated().sum())
```

```python
print(df.dtypes)
```

```python
df['lead_time'] = df['lead_time'].astype(int)
```

```python
df['agent'] = df['agent'].astype(float)
```
