```python
import pandas as pd

# Load the dataset
df = pd.read_csv("Hotel_Reservation_Cleaned.csv")

# Display first 5 rows
print("Original Dataset")
print(df.head())
```

```python
# 1. Filter Rows
# Bookings that were NOT canceled
filtered_df = df[df["is_canceled"] == 0]

print("\nBookings Not Canceled")
print(filtered_df.head())
```

```python
# 2. Select Columns
selected_columns = df[[
    "hotel",
    "city",
    "arrival_date_month",
    "adults",
    "children",
    "adr"
]]

print("\nSelected Columns")
print(selected_columns.head())
```

```python
# 3. Sort Dataset
# Sort by Average Daily Rate (ADR) in descending order
sorted_df = df.sort_values(by="adr", ascending=False)

print("\nSorted Dataset (Highest ADR First)")
print(sorted_df.head())
```

```python
# Save the filtered dataset
filtered_df.to_csv("filtered_dataset.csv", index=False)

print("\nFiltered dataset saved successfully!")
```
