```python
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use("ggplot")
```

```python
# Load Dataset
df = pd.read_csv("C:\\Users\\Pavan\\OneDrive\\Desktop\\Data Science Internship\\Hotel Booking Reservation Dataset.csv")

df.head()
```

```python
# Dashboard Summary
print("HOTEL BOOKING DASHBOARD")
print("-"*40)

print("Total Bookings :", len(df))
print("Total Columns :", len(df.columns))

print("Average Lead Time :", round(df["lead_time"].mean(),2))
print("Average Daily Rate :", round(df["adr"].mean(),2))

print("Cancellation Rate :",
      round(df["is_canceled"].mean()*100,2),"%")

print("Average Stay Duration :",
      round((df["stays_in_weekend_nights"]+
             df["stays_in_week_nights"]).mean(),2),"Days")
```

```python
# Chart 1: Hotel-wise Bookings
plt.figure(figsize=(6,4))

df["hotel"].value_counts().plot(kind="bar", color=["steelblue","orange"])

plt.title("Hotel-wise Bookings")
plt.xlabel("Hotel Type")
plt.ylabel("Bookings")
plt.show()
```

```python
# Chart 2: Booking Cancellation
plt.figure(figsize=(6,4))

df["is_canceled"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    labels=["Not Cancelled","Cancelled"]
)

plt.ylabel("")
plt.title("Booking Cancellation Status")
plt.show()
```

```python
# Chart 3: Monthly Bookings
plt.figure(figsize=(5,5))

df["arrival_date_month"].value_counts().reindex([
    "January","February","March","April","May","June",
    "July","August","September","October","November","December"
]).plot(kind="bar")

plt.title("Monthly Bookings")
plt.xlabel("Month")
plt.ylabel("Bookings")
plt.xticks(rotation=90)
plt.show()
```

```python
# Chart 4: Market Segment
plt.figure(figsize=(5,5))

df["market_segment"].value_counts().plot(kind="bar")

plt.title("Market Segment Distribution")
plt.xlabel("Market Segment")
plt.ylabel("Bookings")
plt.xticks(rotation=45)
plt.show()
```

```python
# Chart 5: Customer Type
plt.figure(figsize=(6,6))

df["customer_type"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.ylabel("")
plt.title("Customer Type Distribution")
plt.show()
```

```python
# Chart 6: Top 10 Cities
plt.figure(figsize=(8,5))

df["city"].value_counts().head(10).plot(kind="barh")

plt.title("Top 10 Cities by Bookings")
plt.xlabel("Bookings")
plt.show()
```

```python
# Chart 8: Reservation Status
plt.figure(figsize=(6,4))

df["reservation_status"].value_counts().plot(kind="bar")

plt.title("Reservation Status")
plt.xlabel("Status")
plt.ylabel("Bookings")
plt.show()
```

```python
# Business Insights
print("BUSINESS INSIGHTS")
print("-"*40)

print("1. Total Bookings :", len(df))

print("2. Most Preferred Hotel :",
      df["hotel"].mode()[0])

print("3. Most Common Customer Type :",
      df["customer_type"].mode()[0])

print("4. Most Common Market Segment :",
      df["market_segment"].mode()[0])

print("5. Average ADR :",
      round(df["adr"].mean(),2))

print("6. Cancellation Rate :",
      round(df["is_canceled"].mean()*100,2),"%")

print("7. Top Booking City :",
      df["city"].mode()[0])

print("8. Average Lead Time :",
      round(df["lead_time"].mean(),2),"Days")
```
